```python
              #القواميس

students = {"name": "ahmed",
           "age": 20,
           "grade": "A"}
#الوصول للقيم
print('Name: ',students["name"])
print('Age: ',students["age"])
print('grade: ',students["grade"])

#تغير قيمة من القاموس
students ["age"] = 22
print ("Age: ",students["age"])
print(students)

#اضافة مفتاح و قيمة للقاموس
students ["city"] = "Rabat"
print(students)

#حدف عنصر من القاموس
del students ["grade"]
print(students)

#المرور على قيم القاموس
x = list(students . keys())  #جمع جميع مفاتيح القاموس في قائمة
i = 0   #بدأ العد من 0
while i < len(x):  #شرط عمل الحلقة
    key = x[i]  #انشاء متغير يمر على مفاتيح القاموس
    print (key,': ',students[key])  #طباعة المفتاح و قيمته
    i += 1   #زياد خطوة في العد  في كل دورة
 
    
    
```

```python
text = "apple banana orange orange apple orange"
character = dict()


for c in text :
    if c not in character :
        character[c] = 1
    else : 
            character[c] += 1
print(character)  

words = text.split() #تحويل النص الى قائمة تحتوي على الكلمات الموجودبه 
dic = {}   #انشاء قاموس فارغ لتخزين المفاتيح و القيم
for x in words :  #المتغير اكس يمر على الكلمات الموجودة في القائمة وورد
    if x not in dic:   #شرط اعطاء قيمة واحد لكل كلمة من القائمة غير موجودة في القاموس
        dic[x] = 1
    else:
         dic[x] += 1  #اضافة واحد لقيمة كل كلمة(مفتاح) موجودة في القاموس
print(words)  #القائمة من كلمات النص 
print(dic)  #القاموس المحدث من الكلمات(المفاتيح) وعددها (القيمة) في النص
print(len(words))    
```

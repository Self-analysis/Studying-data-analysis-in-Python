```python
import pandas as pd

data = {
    "name":['Ali','sara','Omar','Lina','Youssef'],
    "Age":['25','22','33','45','30'],
    "department":['IT','HR','IT','Finance','IT'],
    "salary":[3200,2500,4500,4000,5000]
}

df = pd.DataFrame(data)    #تحويل القاموس الى جدول
df.to_csv("data.csv",index=False)    #تحويل القاموس الى ملف و تخزينة خارج الكود
print(df)
```

```python
#انشاء ملف يحتوي على 10000 صف

import pandas as pd

name = []
age = []
cities = []

for i in range(10000):
    name.append("person_" + str(i))
    age.append(20 + (i % 30))    #حصر الاعمار بين 20 و 49
    cities.append("city_" + str(i % 5))  #حصر المدن بين 0 و 4

data = {
    "Name": name,
    "Age": age,
    "city": cities
}    

df = pd.DataFrame(data)
df.to_csv("data2.csv",index=False)
print(df.head())
```

```python
   #انشاء ملف يحتوي على مليون صف

import pandas as pd

n = 1000000

data = {
    'Name': ['person_'+str(i) for i in range (n) ],
    'Age': [20 + (i % 30) for i in range (n)],
    'City': ['City_' + str(i % 5) for i in range (n)]
}

df = pd.DataFrame(data)
df.to_csv('milion.csv',index=False)
print(df.head())
```

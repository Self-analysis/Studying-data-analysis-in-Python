```python
      #القوائم المتداخلة 

sales = [["ahmed", "phone", '500 dh'],
        ["sara", "laptop", "900 dh"],
         ["karam","smartwatch", "12 dh "]
        ]
     #الوصول للعناصر

print(sales[0]) #طباعة الصف بكامله
print('sara:',sales[1][1],sales[1][2])  #الوصول لعنصر في المصفوفة[رقم الصف][رقم العمود]

```

```python
      #حساب مجموع المبيعات

sales = [
    [12,34,56],  #فبراير
    [54,32,11], #مارس
    [90,7,34]  #ابريل
    ]
Total_1 = sum(sales[0])  #مجموع الصف الاول
Total_2 = sum(sales[1])
Total_3 = sum(sales[2])

print('Total 1:', Total_1 ) 
print('Total 2:',Total_2)
print('Total 3:',Total_3)


   # the total using loop for

for i in range(len(sales)):
    total = sum(sales[i])
    print(i+1,': ',total)
    i += 1


```

```python

grade = [[12,34,23],
        [6,35,23],
        [11,21,89]]
i = 1
for student in grade :
    avg = sum(student) / len(student)
    print('average student',i,':',avg)
    i+=1
```

```python
            # Tuples المجموعات الغير قابلة للتغيير

t = (12,34,56)
t1 = (2,4)
print(t[0])  #الوصول للعناصر
print(len(t))  #طول المجموعة
print(t + t1)  #اضافة مجموعة الى اخرى

for x in t:   ##التكرار
    print (x)

print(t1 * 3)    #التكرار الضربي
```

```python
point = ((12,34,43),
        (6,56,11),
        (10,4,62))
for x , y,z in point:  #تقسيم عناصر المجموعة على ثلات اعمدة
    print('x = ',x,' ', 'y = ',y,' ', 'z = ',z)
```

```python
def min_max(numbers):
    return min(numbers),max(numbers) #إرجاع أكثر من قيمة من دالة واحدة
result = min_max([23,45,67,1])    
print(result)  #النتيجة عبارة عن مجموعة tuple
```

```python
numbers = (22,34,67,89,11,21)
print("maximum: ",max(numbers))
avg = sum(numbers) / len(numbers)
print('average: ',avg)
```

```python
def max_avg (data):
    maximum = max(data)
    average = sum(data) / len(data)
    return maximum , average
numbers = max_avg([22,34,67,89,11,21])
print(numbers)
```

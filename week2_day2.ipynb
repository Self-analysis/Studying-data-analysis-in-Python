```python
# function and loop

#استعمال الدالة داخل الحلقة

def simo(x):          #تعريف الدالة
    return x * x      #ارجاع مهمة الدالة

data = [2,3,4,5]
for n in data:
    square = simo(n)
    print("Square ",n," = ",square)   
```

```python
#استعمال الحلقة داخل الدالة

def adding (numbers):   #تعريف الدالة
    total = 0           #متغير لتخزين النتيجة
    for x in numbers:   #استعمال الحلقة لتكرار العملية داخل الدالة
        total = total + x   #جمع القيم و تخزينها
    return total        #ارجاع عمل الدالة

data = [23,45,67,89,]
result = adding (data)
print("adding = ",result)
```

```python
def avg (numbers):   #تعريف الدالة 
    total = 0        #متغير لتخزين المجموع
    for x in numbers:#حلقة لتكرار العملية في جميع القيم
        total += x   #تخزين و جمع القيم
    return total / len(numbers) #مهمة الدالة حساب المتوسط
data = [45,67,89,32,12]    
print("Average = ",avg(data))
```

```python
def value (numbers):
    value_max = max(numbers)  #اجاد اكبر قيمة 
    return value_max
data = [34,55,554,123]   
print("Max value: ",value(data))
```

```python
def value (numbers):
    max_value = numbers[0]  #متغير يخزن اكبر قيمة و نعتبر الرقم الاول في القائمة اكبر قيمة كبداية 
    for x in numbers:  
        if x > max_value:  #شرط تخزين اكبر قيمة
            max_value = x
    return max_value

data = [23,45,67,98]
print("Max value: ",value(data))
```

```python


def print_data (data):    #تعريف الدالة
    i = 0                 #متغير لتخزين قيمة إ
    while i < len(data):  #شرط استمرار الحلقة
        print(data[i])    #طباعة قيم القائمة بالترتيب 
        i += 1            #زيادة خطوة في ترتيب القائمة 
    

numbers = [34,21,67,89,11]      
print_data(numbers)
```

```python
def greater (data):  #تعريف الدالة
    count = 0        #متغير عبار عن عداد لتخزين عدد الاعداد الاكبر من 10
    i = 0
    while i < len(data):   #شرط استمرار الحلقة
        if data[i] > 10 :  #شرط تخزين عدد الاعداد الاكبر من 10
            count += 1     #زيادة واحد للعداد ادا تحقق الشرط
        i += 1             #الانتقال من قيمة الى قيمة اخرى في ترتيب القائمة بواحد 
    return count           #ارجاع قيمة المتغير كونت اي العداد
        
numbers = [2,44,6,78,12,9]    
print("The number of numbers greater than 10 is : ",greater(numbers))
    
```

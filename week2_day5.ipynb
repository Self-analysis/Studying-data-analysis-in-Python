```python

import pandas as pd

data = {
    'Name': ['karam','asad','charif','sara'],
    'Age': [22,44,30,25],
    'Department': ['IT','IT','FI','HR'],
    'City': ['rabat','casablanca','rabat','agadir'],
    'Salary': [4000,3500,6000,9000,]
}

df = pd.DataFrame(data)
df.to_csv('ouvrier.csv',index=False)

#df = pd.read_csv("ouvrier.csv",engine="c")  #قراءة ملف خارجي
#print(df.head())  #طباعة اول5 صفوف من الملف
#print(df.info())  #الاطلاع على معلومات عن الملف

avg_salary = df['Salary'].mean()
print('Average Salary =',avg_salary)

#حساب متوسط الراتب لكل قسم

print('average salary by department:')
avg_by_dept = df.groupby("Department")['Salary'].mean()
for dept,avg in avg_by_dept.items():
    print(dept,':',avg)
#تخزين الناتج في ملف جديد
avg_by_dept.to_csv('avg_by_dept.csv')

highest_salary = df['Salary'].max()
print('Highest Salary =',highest_salary)

minimum_salary = df['Salary'].min()
print('Minimum Salry =',minimum_salary )

#حساب عدد التكرار في كل قسم
depart_count = df['Department'].value_counts() 
print(depart_count)

#تنظيم دالة حساب التكرار لتفادي معلومات اضافية
print('# Number of Employees in esch Department:')
for dept,cont in df['Department'].value_counts().items(): 
    #هذه الدالة تحول النتيجة إلى أزواج items()(key, value)
     print(dept,':',cont)

#استخراج الموظفين الذين راتبهم أكبر من 4500
high_salary = df[df['Salary']  > 4500]
print('# High Salary:')
print(high_salary)
```

```python
#حساب المتوسط
def average(x):
    avg = sum(x) / len(x)
    return avg

data = [23,56,78,9,0]
print('average: ',average(data))
```

```python


def min_max(numbers):  #تعريف الدالة
    result = min(numbers) ,max(numbers)  #المهمة المخزنة في الدالة
    return result  #عملية ارجاع الدالة عبارة عن مجموعة 
    

data = [34,67,89,1,90]
x = min_max(data)  #تخزين النتيجة في متغير النتيجة عبارة عن مجموعة
print(x)
 #تحديد العنصر المراد ارجاعه من المجموعة
print('minimum: ',x[0],' /maximum: ',x[1]) 
```

```python
        #تنظيف البيانات

def remove_negative(numbers):
    clear_data = []
    for x in numbers:
        if x >= 0:
            clear_data.append(x)
    return clear_data
data =[34,56,-12,2,-90]    
print(remove_negative(data))
```

```python
     # Pandas استخدام الدوال مع  

import pandas as pd

data = {
    'age': [34,26,50],
    'salary': [3000,5000,8000]
}
df = pd.DataFrame(data) #تحويل القاموس الى جدول و تخزينه في متغير
print("Highest Salary: ",df['salary'].max())
print("Minimum Salary: ",df['salary'].min())
print("average Salaries: ",df['salary'].mean())
```

```python
import pandas as pd 
def analys_data (numbers):
    return {
        "minimum: ": min(numbers),
        "maximum: ": max(numbers),
        "average: ": sum(numbers) / len(numbers)
    }

data = [23,45,90,76,88]    
print(analys_data(data))
```

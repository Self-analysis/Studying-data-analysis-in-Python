```python
# while الحلقة 

while True:  #الحلقة تعمل بلا نهاية
    x = input("name :")  #تخزين الادخال في متغير
    if x == 'stop':  #شرط توقف الحلقة عن العمل
        break
    print('your name :',x)  #طباعة المدخلات   
```

```python
total = 0  #انشاء متغير فارغ

while (True):  #حلقة بدون نهاية
    #انشاء متغير لتخزين المدخلات
    x = input('enter a number or type done to exit :')
    if x == 'done':   #شرط توقف الحلقة
        break
    num = float(x)  #تحويل المدخلات الى عدد عشري
    #استعمال المتغير لتخزين مجموع الاعداد
    total = total + num  
    print('Current Total :',total)  #طباعة المجموع الحالي
print('Final Total :',total)  #  طباعة المجموع النهائي عند الخروج من الحلقة   
```

```python
#حساب متوسط المجموع بدون قائمة
total = 0
num_entries = 0
while (True):
    x = input('enter number or done to exit :')
    if x == 'done':
        break
        #تخزين مجموع الاعداد المدخلة
    total = total + float(x)
    #حساب عدد المدخلات
    num_entries = num_entries + 1   
avg = total / num_entries 
print("Average :",avg)
    
```

```python
#حساب متوسط مجموع الاعداد مع بعض الشروط المنطقية
total = 0
count = 0
while True :
    x = input("enter number or done to exit : ")
    if x == 'done':
        break
    value = float(x)
    total += value
    count += 1 #زيادة عدد المدخلات ب واحد في كل مرة يتم فيها الادخال
if count > 0 :
    avg = total / count
    print("Average :",avg)    
else :
    print("no numbers was entred")
```

```python
# list and function

list = [12,34,6,54]
print(list[0])#الوصول الى عنصر في القائمة
print(list.index(12))  #الوصول الى موضع عنصر في القائمة
list.append(100)  #اضافة عنصر الى القائمة
print(list)

list[1] = 50  #تغيير عنصر في القائمة
print(list)

list.insert(1, 15)  #اضافة عنصر في مكان محدد 1 
print(list)

list.pop()  #حدف العنصر الاخير من القائمة لانه لم يتم تحديد اي موضع الحدف 
x = list.pop() #الاحتفاظ بالعنصر المحدوف داخل متغير
print(list)

print(x)
del list[0]  #حدف نهائي بموضع العنصر
print(list)

list.remove(6)  #حدف بقيمة العنصر
print(list)

print("Number of elements: ",len(list))  #طباعة عدد عناصر القائمة 
print("Highest value: ",max(list))  #طباعة اعلى قيمة في القائمة
print("Lowest value: ",min(list))  #طباعة ادنى قيمة في القائمة
print("Total value: ",sum(list))  #طباعة مجموع القيم في القائمة
print("Average: ",sum(list) / len(list))  #طباعة متوسط القيم في القائمة

# forالتكرار بحلق 
mun = [1,2,3,4]
for x in mun:
    print('تكرار x',x)

numbers = [1,2,3,4,5,6]
squared = [x**2 for x in numbers]  #انشاء قائمة من مربع عناصر القائمة الاولى 
print(squared,'مربع اكس')

# الطريقة السهل و المفضلة ان لم يكن هناك تعقيد في الحساب
even_numbers = [y for y in numbers if y % 2 == 0]  
print("even numbers =", even_numbers)

#نستخدمها  للعمليات المعقدة for
even2 = []
for num in numbers :
    if num % 2 == 0:
        even2.append(num)
print("even numbers 2: ",even2)
    
```

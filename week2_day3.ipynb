```python
     #دالة حساب المجموع

def adding (data):
    total = 0
    for x in data:
        total += x
    return total

numbers = [45,56,23,11]
print("Total = ",adding(numbers))
```

```python
     #دالة حساب المتوسط

def avg (data):
    total = 0
    for x in data:
        total += x
    return total / len(data)   

numbers = [89,77,56,23]
print("Average = ",avg(numbers))
    
```

```python
      #دالة إيجاد أكبر قيمة

def highest_value (data):
    max_value = data[0]
    for x in data:
        if x > max_value:
            max_value = x
    return max_value

numbers = [1,23,89,65]
print("Highest value = ",highest_value(numbers))
```

```python
      #دالة عدّ القيم

def counter (data):
    coun = 0
    for _ in data:
        coun += 1
    return coun

numbers =[2,3,6,5]
print("Number of Values = ",counter(numbers))
```

```python
    #دالة عدّ القيم الأكبر من رقم معين

def greater (data):
    great = 0
    for x in data:
        if x > 10:
            great += 1
    return great

numbers = [11,1,34,9,10]  
print("Number of numbers greater than 10 = ",greater(numbers))
```

```python
     #دالة استخراج القيم الأكبر من رقم

def greater (data):
    great = []
    for x in data:
        if x > 10:
            great.append(x)
    return great
numbers = [11,1,34,9,10] 
print("Number of Numbers greater than 10 : ",greater(numbers))
```

```python
     #دالة جمع الأعداد الزوجية فقط

def sum_even_number (data):
    total = 0
    for x in data:
        if x % 2 == 0 :
            total += x
    return total

numbers = [1,2,5,3,6,4,8,9]
print('Total even numbers = ',sum_even_number(numbers))
```

```python
    #دالة البحث عن قيمة داخل البيانات
def search (data,target):
    for x in data:
        if x >= target:
            return True
    return False
target = 90
sales = [100]
print(search(sales,target))
```
